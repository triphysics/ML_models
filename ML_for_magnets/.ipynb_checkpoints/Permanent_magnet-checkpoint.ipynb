# XGBoost vs ANN: train, evaluate, SHAP & partial dependence

This notebook trains XGBoost and a small ANN, produces parity plots, ANN diagnostics, SHAP explanations (KernelExplainer) and partial dependence plots. Run in Colab or locally.

```python
import os, math, random, joblib, warnings
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from sklearn.model_selection import train_test_split, KFold
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.inspection import PartialDependenceDisplay
import xgboost as xgb
import torch, torch.nn as nn, torch.optim as optim
import shap

warnings.filterwarnings('ignore')
SEED = 42
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", DEVICE)

# -------------------- helper functions --------------------

def load_or_parse_table(csv_path="Table_XIII.csv", raw_path="Table_XIII_raw.txt"):
    if os.path.exists(csv_path):
        df = pd.read_csv(csv_path)
        print(f"Loaded parsed CSV: {csv_path} ({len(df)} rows)")
        return df
    if not os.path.exists(raw_path):
        raise FileNotFoundError("Provide Table_XIII_full_parsed.csv or Table_XIII_raw.txt in the working directory.")
    raw = open(raw_path, "r", encoding="utf-8").read()
    import re
    m = re.search(r"Material\s+Ms\s+\(MA/m\)\s+Aex\s+\(pJ/m\)\s+Ku\s+\(MJ/m3\)\s+Hc\s+\(T\)", raw, re.IGNORECASE)
    start = m.end() if m else 0
    body = raw[start:]
    lines = [ln.strip() for ln in body.splitlines() if ln.strip()]
    rows = []
    for line in lines:
        # try robust extraction of final four numeric columns
        m2 = re.search(r"(-?\d+\.\d+)\s+(-?\d+\.\d+)\s+(-?\d+\.\d+)\s+(-?\d+\.\d+)$", line)
        if m2:
            nums = m2.groups()
            mat = line[:m2.start()].strip()
            rows.append((mat, float(nums[0]), float(nums[1]), float(nums[2]), float(nums[3])))
        else:
            m3 = re.findall(r"(-?\d+\.\d+)", line)
            if len(m3) >= 4:
                nums = m3[-4:]
                idx = line.rfind(nums[0])
                mat = (line[:idx].strip() if idx!=-1 else line)
                rows.append((mat, float(nums[0]), float(nums[1]), float(nums[2]), float(nums[3])))
    df = pd.DataFrame(rows, columns=["Material","Ms (MA/m)","Aex (pJ/m)","Ku (MJ/m3)","Hc (T)"])
    df = df.drop_duplicates(subset=["Material"]).reset_index(drop=True)
    print(f"Parsed from raw text: {len(df)} rows")
    return df


def prepare_features(df, extra_feature_engineering=True):
    df = df.copy()
    # Keep canonical numeric columns
    cols = ["Ms (MA/m)", "Aex (pJ/m)", "Ku (MJ/m3)", "Hc (T)"]
    for c in cols:
        if c not in df.columns:
            raise KeyError(f"Required column {c} missing from input dataframe")
    df = df[cols + [c for c in df.columns if c not in cols and c=="Material"]]
    df = df[df["Hc (T)"].notna()].reset_index(drop=True)
    df["Ha (T)"] = df.apply(lambda r: 2.0 * r["Ku (MJ/m3)"] / r["Ms (MA/m)"] if pd.notna(r["Ku (MJ/m3)"]) and pd.notna(r["Ms (MA/m)"]) and r["Ms (MA/m)"]!=0 else np.nan, axis=1)
    df["log_Hc"] = np.log(df["Hc (T)"])
    df["log_Ku"] = np.log(df["Ku (MJ/m3)"])
    df["log_Ms"] = np.log(df["Ms (MA/m)"])
    return df


def metrics(y_true, y_pred, prefix=""):
    rmse = math.sqrt(mean_squared_error(y_true, y_pred))
    mae = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    print(f"{prefix} RMSE: {rmse:.4f}, MAE: {mae:.4f}, R2: {r2:.4f}")
    return {"rmse":rmse, "mae":mae, "r2":r2}


def parity_plot(y_true, y_pred, title="Parity plot"):
    fig, ax = plt.subplots(figsize=(5,5))
    ax.scatter(y_true, y_pred, alpha=0.8)
    mn = min(np.nanmin(y_true), np.nanmin(y_pred))
    mx = max(np.nanmax(y_true), np.nanmax(y_pred))
    ax.plot([mn, mx], [mn, mx], 'k--', linewidth=1)
    ax.set_xlabel('Actual Hc (T)')
    ax.set_ylabel('Predicted Hc (T)')
    ax.set_title(title)
    return fig, ax
```

```python
# -------------------- Load and prepare data --------------------
# Place either Table_XIII_full_parsed.csv or Table_XIII_raw.txt in the working directory
try:
    df = load_or_parse_table(csv_path="Table_XIII.csv", raw_path="Table_XIII_raw.txt")
except Exception as e:
    raise RuntimeError(f"Could not load/parse Table XIII: {e}")

# Feature engineering
df = prepare_features(df)
print("Total parsed rows:", len(df))
# Drop rows missing core features
df = df.dropna(subset=["Ms (MA/m)", "Aex (pJ/m)", "Ku (MJ/m3)", "Hc (T)", "Ha (T)", "log_Ku", "log_Ms"]).reset_index(drop=True)
print("Usable rows after dropping incomplete:", len(df))

# Choose baseline features
FEATURES = ["Ms (MA/m)", "Aex (pJ/m)", "Ku (MJ/m3)", "Ha (T)", "log_Ku", "log_Ms"]
X = df[FEATURES].values.astype(np.float32)
y = df["Hc (T)"].values.astype(np.float32)
mask = y > 0
X = X[mask]; y = y[mask]

# Train-test split (70:30)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=SEED)
scaler = StandardScaler(); X_train_s = scaler.fit_transform(X_train); X_test_s = scaler.transform(X_test)
```

```python
# -------------------- XGBoost --------------------
print('\nTraining XGBoost...')
xgb_model = xgb.XGBRegressor(n_estimators=1200, max_depth=5, learning_rate=0.007, objective='reg:squarederror', random_state=SEED)
xgb_model.fit(X_train, y_train)
y_pred_xgb = xgb_model.predict(X_test)
metrics(y_test, y_pred_xgb, prefix='XGBoost test')

# Feature importance
print('Feature importances (XGBoost):')
for name, val in zip(FEATURES, xgb_model.feature_importances_):
    print(f"  {name}: {val:.4f}")

# Parity and residuals for XGB
fig, ax = plt.subplots(1,2, figsize=(12,5))
ax0 = ax[0]; ax1 = ax[1]
ax0.scatter(y_test, y_pred_xgb, alpha=0.8); ax0.plot([y_test.min(), y_test.max()],[y_test.min(), y_test.max()],'k--'); ax0.set_title('XGBoost parity'); ax0.set_xlabel('Actual Hc'); ax0.set_ylabel('Predicted Hc')
ax1.hist(y_test - y_pred_xgb, bins=20); ax1.set_title('XGBoost residuals (Actual - Predicted)')
plt.tight_layout(); plt.show()
```

```python
# -------------------- Small ANN (PyTorch) --------------------
print('\nTraining Small ANN (PyTorch)...')
class SmallANN(nn.Module):
    def __init__(self, input_dim, hidden_dims=[64,32,16], dropout=0.1):
        super().__init__()
        layers = []
        in_d = input_dim
        for h in hidden_dims:
            layers.append(nn.Linear(in_d, h)); layers.append(nn.LeakyReLU(0.01))
            if dropout and dropout>0: layers.append(nn.Dropout(dropout))
            in_d = h
        layers.append(nn.Linear(in_d, 1))
        self.net = nn.Sequential(*layers)
    def forward(self, x): return self.net(x)

# Tensors
X_tr_t = torch.tensor(X_train_s, dtype=torch.float32).to(DEVICE)
X_te_t = torch.tensor(X_test_s, dtype=torch.float32).to(DEVICE)
y_tr_log = np.log(y_train); y_te_log = np.log(y_test)
y_tr_t = torch.tensor(y_tr_log.reshape(-1,1), dtype=torch.float32).to(DEVICE)
y_te_t = torch.tensor(y_te_log.reshape(-1,1), dtype=torch.float32).to(DEVICE)

model = SmallANN(input_dim=X_train_s.shape[1], hidden_dims=[64,32,16], dropout=0.1).to(DEVICE)
opt = optim.Adam(model.parameters(), lr=0.007, weight_decay=1e-4)
crit = nn.MSELoss()

batch_size = 24; epochs = 800; patience = 80
best_val = float('inf'); epochs_no_improve = 0; best_state = None
train_losses = []; val_losses = []

for epoch in range(1, epochs+1):
    model.train()
    perm = torch.randperm(X_tr_t.size(0), device=DEVICE)
    epoch_losses = []
    for i in range(0, X_tr_t.size(0), batch_size):
        idx = perm[i:i+batch_size]
        xb = X_tr_t[idx]; yb = y_tr_t[idx]
        pred = model(xb); loss = crit(pred, yb)
        opt.zero_grad(); loss.backward(); opt.step()
        epoch_losses.append(loss.item())
    # validation
    model.eval()
    with torch.no_grad():
        val_pred_log = model(X_te_t).cpu().numpy().flatten()
        val_loss = mean_squared_error(y_te_log, val_pred_log)
    train_losses.append(np.mean(epoch_losses)); val_losses.append(val_loss)
    if val_loss < best_val - 1e-9:
        best_val = val_loss; epochs_no_improve = 0; best_state = {k: v.cpu().clone() for k,v in model.state_dict().items()}
    else:
        epochs_no_improve += 1
        if epochs_no_improve >= patience:
            print(f"Early stopping at epoch {epoch} (best val MSE ln: {best_val:.6f})")
            break
    if epoch % 100 == 0 or epoch==1:
        print(f"Epoch {epoch}/{epochs} - train MSE (ln): {np.mean(epoch_losses):.5f} - val MSE (ln): {val_loss:.5f})")

# load best
if best_state is not None:
    model.load_state_dict(best_state)
```

```python
# Evaluate ANN on test set
model.eval()
with torch.no_grad():
    y_test_pred_log = model(X_te_t).cpu().numpy().flatten()
    y_test_pred = np.exp(y_test_pred_log)
metrics(y_test, y_test_pred, prefix='ANN test')

# ANN parity, residuals, learning curve
fig, ax = plt.subplots(1,3, figsize=(15,4))
ax[0].scatter(y_test, y_test_pred, alpha=0.8); ax[0].plot([y_test.min(), y_test.max()],[y_test.min(), y_test.max()],'k--'); ax[0].set_title('ANN parity'); ax[0].set_xlabel('Actual Hc'); ax[0].set_ylabel('Predicted Hc')
ax[1].hist(y_test - y_test_pred, bins=20); ax[1].set_title('ANN residuals')
ax[2].plot(train_losses, label='train MSE (ln)'); ax[2].plot(val_losses, label='val MSE (ln)'); ax[2].set_title('ANN learning curve (MSE ln)'); ax[2].legend()
plt.tight_layout(); plt.show()
```

```python
# -------------------- SHAP explanations for ANN --------------------
print('\nRunning SHAP (KernelExplainer) for ANN — this may take time, uses a small background sample by default.')
# wrapper function: input in original feature scale -> returns predictions (Hc)
def ann_predict_numpy(X_np):
    Xs = scaler.transform(X_np)
    Xt = torch.tensor(Xs, dtype=torch.float32).to(DEVICE)
    with torch.no_grad():
        out = model(Xt).cpu().numpy().flatten()
    return np.exp(out)

# choose small background to keep runtime reasonable
bg_size = min(50, X_train.shape[0])
background = X_train[np.random.choice(X_train.shape[0], bg_size, replace=False)]
explainer = shap.KernelExplainer(ann_predict_numpy, background)
# explain a subset of test points
explain_size = min(20, X_test.shape[0])
to_explain = X_test[np.random.choice(X_test.shape[0], explain_size, replace=False)]
shap_values = explainer.shap_values(to_explain, nsamples=200)

# SHAP summary
print('\nSHAP summary plot (ANN)')
shap.summary_plot(shap_values, to_explain, feature_names=FEATURES, show=True)

# SHAP dependence plots for each feature (limited set)
for i,fname in enumerate(FEATURES):
    try:
        shap.dependence_plot(i, shap_values, to_explain, feature_names=FEATURES, show=True)
    except Exception as e:
        print('SHAP dependence plot failed for', fname, e)
```

```python
# -------------------- Save models & scaler --------------------
xgb_model.save_model('xgb_model.json')
torch.save({'model_state_dict': model.state_dict()}, 'ann_model.pt')
joblib.dump(scaler, 'scaler.joblib')
print('\nSaved models: xgb_model.json, ann_model.pt, scaler.joblib')

# End of file
```
